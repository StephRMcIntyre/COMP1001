### Accessing elements 

To access elements in a list we use `[]` operator.  In Python, list indexing starts a 0. This means the first element has index 0, second has index 1, and so on. Now, let us get the third meal in the `meals` list.

```python
print(meals[2])
```

The only thing we need to make sure of if that we use an integer when indexing. **List indexing syntax:** `some_list[index]`.

One natural question that may arise now is: if the index must be an integer, then can we use negative indices?  The answer is yes. Before running the example below, what do you think the output is going to be if we change `meals[2]` to `meals[-2]`?

### Slicing a list

You'll often find that you need portions of a list, this is referred to as a slice. The slicing syntax is as follows: 
`some_list[start:stop]`,
`some_list` is the list you want to take a slice from, the slice will begin at index `start` , end at the index preceding `stop`. Slicing **does not** include the `stop` index. If you do not define `start`, then you will see behaviour of `start=0`. If you do not define `stop`, then you will see behaviour of `stop=len(some_list)-1`. Below, we do not define `start` and `stop` when slicing the `meals` list. Thus, you should expect the entire list to be printed. Run the example below to verify this conclusion.


```python
print(meals[:])
```

To get everything up to but not including the fifth index from the `course_evaluation` list, 

```python
print(course_evaluation[:5])
```

To get everything after the second index from the `course_evaluation` list,

```python
print(course_evaluation[2:])
```

To get everything up to between the second and fifth indices from the `course_evaluation` list,

```python
print(course_evaluation[2:5])
```

You may have noticed that the slice takes every element between `start` and `stop`. There may be times you want to get every second, third, or fourth index. This is done by defining the `step` value. The syntax is as follows: `some_list[start:stop:step]`, `some_list`, `start` , and `stop` are the same as before. `step` defines the step size. Let us get all the string data from the `course_evaluation` list. This corresponds to getting only the even indices starting at 0

```python
print(course_evaluation[::2])
```

You are not limited to slicing lists, you can also slice a strings.

```python
string = "COMP 1001 is my favorite course"
# get every second character from the string
print(string[::2])
```

### Arithmetic  operators


There are a few arithmetic operators that you can use with lists, `+` and `*`.

Adding of two lists, `list_1 + list_2`,  will concatenate  two lists together.

```python
print(course_evaluation + meals)
```

The `list_1 * some_integer_number` can be thought of as creating `some_integer_number` of copies of `list_1`, then adding them all together.

```python
print(meals*2)
```

## List methods

### Append items

We use `append()` to add items to the end of our lists. The code `some_list.append(item)` will add `item` to the end of `some_list`. Below is list of some fruits, we will append pineapple to our list.

```python
fruits = ["apple", "oranges","banana"]
print("fruits in our list before appending: ",fruits)
fruits.append("pineapple") # add pineapple to the list
print("fruits in our list after appending: ",fruits)
```

### Insert items

When you have a specific place we want to insert our new item, you use `insert()`. `some_list.insert(index,item)` will insert `item` before `index` in `some_list`.  

```python
print("fruits in our list before instert: ",fruits)
fruits.insert(1,"kiwi") # insert kiwi before oranges
print("fruits in our list after instert: ",fruits)
```

### Remove data

We can remove items from our list using `remove()`. `some_list.remove(item)` will remove the first instance of `item` from the `some_list`. 

```python
fruit_returned = fruits.remove("pineapple") # remove pineappe from our list
print("fruits list: ",fruits)
print("value returned by remove(): ",fruit_returned)
```

If you try to remove an item that does not appear in the list, you will get an error telling you that the item is not in the list.

```python
fruits.remove("pineapple") # try to remove pineapple a second time
```

### Pop item

The downside to using `remove()`, is that we do not get the item back. This means, if you use `some_list.remove(item)`, `item` is no longer accessable to you. Instead, `pop()` will remove the item and return it to you. `some_list.pop()` will remove items from the end of the list. 

```python
print("fruits before pop():",fruits) 
fruit_returned = fruits.pop()
print("fruits after pop():",fruits)
print("fruit returned by pop():",fruit_returned)
```

The default action for `pop()` is to remove items from the end of the list. You can change this by specifying  the index. `some_list.pop(index)` will remove and return the item at `index` from `some_list`.

```python
print("fruits before pop(1):",fruits)
fruit_returned = fruits.pop(1) # remove fruit at index 1, should be kiwi
print("fruits after pop(1):",fruits)
print("fruits returned by pop():",fruit_returned)
```

### Index

If you want to know the index of an `item` in a given list, use the `index()`. `some_list.index(item)` you will get the first index of `item` from `some_list`.

```python
print("oranges is at index:",fruits.index("oranges"))
```

If the `item` you want does not appear in the list you will get an error stating the item is not in the list

```python
# we misspelling of oranges 
print("oranges is at index:",fruits.index("orange"))
```

## Summary

 - Lists are created using `[]`. 
 - `[]` is used to access individual elements within the list.
 - `len(some_list)` will return the length of the list `some_list`
 - `some_list[start:stop:step]` slices `some_list` begining at `start` going up to but not including `stop`, with a step size of `step`
 - `list_1 + list_2` concatinates two lists together
 - `list_1 * some_integer_number` is just repeated addiing `list_1` to itself `some_integer_number` of times
 - `some_list.append(item)` will add `item` to the end of `some_list`
 - `some_list.insert(index,item)` will insert `item` before `index`
 - `some_list.remove(item)` will remove the first instance of `item` from the `some_list`. 
 - `some_list.pop()` will remove the item at the end of `some_list`
 - `some_list.pop(index)` will remove the item at `index`
 - `some_list.index(item)` will return the first index of `item`
