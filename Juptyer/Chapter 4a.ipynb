### While loop

A `while` loop will repeat the loop body until the `loop condition` is `False`. In general, the `while` loop syntax is as follow,
 ```
     while (loop condition):
         loop body code executed while loop condition is True
     code after loop condition is False
 ```

The `loop condition` is written with the same syntax as a the `if` statement seen in chapter 3. The loop body code must indented, indicating the code is part of the loop.  

Run the example below and observe the output.

````python
counter = 0 # initialize variable used in the loop condition
while (counter < 5):
    print("In the loop counter =",counter) 
    counter = counter + 1
print("Outside the loop counter =",counter)
````

We saw that until `counter = 5`, the loop condition `counter < 5` is `True` causing the loop body to be repeated. Every time the loop body is repeated we call this an **iteration**. When `counter = 5`, the loop condition `counter < 5` is `False` causing the loop body to terminate. We will also trace the code (shown below) to help illustatre the while loop process.

## Sentinel values


A **sentinel value** is a value that denotes the end of a data set, but it is not part of the data set. Sentinel values uises include but are not limited to:
 - using a negative number to indicate the end of a sequence of positive numbers 
 - using a special character or value to denote the last item


The example below is a basic cash register that keeps a running total. When the cashier is finished inputing all items, they enter the word 'done'. Afterwards, we calculate the total with taxes and print that value to the customer. Play around with the example below, 

````python
done = False # define sentinel valuie
total = 0.0 # initalize running total
while not done: 
    cost = input("Please enter the cost of the item or the word 'done':") # obtain user input
    if cost.lower() == 'done':  # check to see if we are done 
        done = True # set done to True
    else: # update the total 
        total = total + float(cost)
        print("The current total is $%1.2f" % total)
print("Your total after taxes comes to $%1.2f " % (total*1.15)) # output total
````

## Common Errors 


### Infinite loops
Becareful, when writing the loop condition.  It is easy to get into situations where the loop never terminates, this is known as an infinite loop. 

Below we have an example of an infinite loop, identify the error, fix the error, then run the code. 

````python
iteration = 0 
MAX_ITERATION = 10
# this loop is an infinite loop
# fix the error that is causing it to run forever
# you can press the stop button to terminate the cell
while iteration < MAX_ITERATION:
    print("Working hard iterating some code.")
print("Phew, I'm finally done. Only took %d iterations" % iterations)
````

Although, the above example is fairly trivial. It is not uncommon for you to run into this sort of mistake in out in the wild. You need to make sure that you are update the variable that is controling the loop. 

### Incorrect logic

We need to remember that the loop body is run when the loop condition is `True`.  Suppose, we set `iteration = 1` and `MAX_ITERATION=10`. If you wrote `iteration > MAX_ITERATION` instead of `iteration < MAX_ITERATION`. Then the loop body would never execute. 

If you have a loop that isn't executing, then you should check that loop condition is correct. 

### Off by one error

An off by one error is when the loop either iterates one too many or too few times. From our above example, after 10 iterations the loop condition should be false. There are two places we can identify where an off by one error may occur.
 - Do we start `iteration` at 0 or 1? 
 - Do we test for `iteration < MAX_ITERATION` or `iteration <= MAX_ITERATION`

Let us break this down into the four possible cases with `MAX_ITERATION=10` for all of them. That is, we want to loop to execute at most 10 times. The code below will print out the number of times the loop body was executed after the loop is done.

> Case 1:`iteration=0` and `iteration < MAX_ITERATION`

````python
iteration = 0 
MAX_ITERATION = 10 # maximum number of times the loop should run (inclusive)
loop_body_executions = 0 # keep track of the number of loop body executions
while iteration < MAX_ITERATION:
    iteration = iteration +1
    loop_body_executions =  loop_body_executions + 1
print("Loop executed %d times" % loop_body_executions)
````

> Case 2: `iteration=1` and `iteration < MAX_ITERATION` 

````python
iteration = 1 
MAX_ITERATION = 10 # maximum number of times the loop should run (inclusive)
loop_body_executions = 0 # keep track of the number of loop body executions
while iteration < MAX_ITERATION:
    iteration = iteration +1
    loop_body_executions =  loop_body_executions + 1

print("Loop executed %d times" % loop_body_executions)
````

> Case 3: `iteration=0` and `iteration <= MAX_ITERATION` 


````python
iteration = 0 
MAX_ITERATION = 10 # maximum number of times the loop should run (inclusive)
loop_body_executions = 0 # keep track of the number of loop body executions
while iteration <= MAX_ITERATION:
    iteration = iteration +1
    loop_body_executions =  loop_body_executions + 1
print("Loop executed %d times" % loop_body_executions)
````

> Case 4: `iteration=1` and `iteration <= MAX_ITERATION` 

````python
iteration = 1 
MAX_ITERATION = 10 # maximum number of times the loop should run (inclusive)
loop_body_executions = 0 # keep track of the number of loop body executions
while iteration <= MAX_ITERATION:
    iteration = iteration +1
    loop_body_executions =  loop_body_executions + 1
print("Loop executed %d times" % loop_body_executions)
````

There are two cases that had the desired effect we stated earlier, with case 1 and 4. It's not hard to run into a off by one error. In general, if your while loop has a lower bound of `a` and an upper bound of `b`, then the loop with`a<b` will execute `b-a` times and `a<=b` will execute `b-a+1` times.

### Compute sum of values from user input

To compute the sum from user input we need to:
- Initialize a variable to 0 that stores the current sum
- Get user input
- Detemine sentinal value the signals the end of user input
- Convert the user input from a string to a numeric value
- Increment total with user input 
- Get user input again
- (Optional) Print result 


````python
# initialize running total variable as a float
total = 0.0 

# get user input 
user_input = input("Enter value or leave blank to terminate: ") 

# if user_input is empty terminate loop
while user_input != "": 
    
    # convert user_input to a numeric value
    user_value = float(user_input) 
    
    # add value to running total
    total = total + user_value
    
    # ask for user input again
    user_input = input("Enter value or leave blank to terminate: ") 
    
# print the result
print("The total is:",total)
````

### Prompting user input until a match is found

To prompting user input until a match is found we need to:
- Initialize a boolean variable as our sentinal value
- Define our target value
- Get user input
- Convert the user input from a string to a numeric value
- Check if new user input is larger then current maximum 
    - If above is true, update current maximum to user input
- Get user input again
- (Optional) Print result 

````python
# Initialize a boolean variable as our sentinal value
done = False 

# Define our target value
target = 14



# if user_input is empty terminate loop
while not done: 
    
    # get user input 
    user_input = input("Enter an integer between 0 and 20: ") 
    
    # convert user_input to a numeric value
    user_value = int(user_input) 
    
    # Check if new user input is larger then current maximum 
    if (user_value == target):        
        done = True
        print("Yay, you got it!")
    else:
        print("Oops, wrong answer. Please try again.")

    
````
