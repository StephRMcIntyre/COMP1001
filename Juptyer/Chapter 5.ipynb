### Return

A `return` is used to exit a function and return work done in the function back to the function caller. Then, the program resumes from where the function call was made. By default, if you do not specify a return value, then Python will use the default return value of `None`. 

As an exercise, comment out `return result` from the `inches_to_centimeter` function above. Then, rerun the example and observe the output. 

The `return` statement may return a variable, a value, or the result of an expression. Moreover, there can be more then one return statement. 

Below, we modified the `inches_to_centimeter` function to show how you might create a function with two return statments. The original one is a returns a variable, the second return is in a branch that returns a value. 

````python
def inches_to_centimeter(inch):
    # any negative value should have the function return 0
    if inch < 0: # create a branch
        # return a value
        return 0 
    # one is 2.54 cm
    result = inch*2.48 
    # return a variable
    return result 

inches = 1.0
print("%1.2f inches is %1.2f cenimeters" %(inches,inches_to_centimeter(inches)))

inches = -1.0
print("%1.2f inches is %1.2f cenimeters" %(inches,inches_to_centimeter(inches)))
````

It should be noted that, with the above example, we want to make sure that the branch we made with the if statement should have a return statement. The compiler does **not** complain if your branch does not have a return. This might lead to run-time errors if the function caller requires the result. 

The next two cells demonstrate how you might structure a function with multiple branches. In both of them, all the condionals are handled.

````python
#
# Have a return in each branch
#
def inches_to_centimeter(inch):
    if inch < 0: 
        return 0 
    else:
        result = inch*2.48 
        return result 

inches = 1.0
print("%1.2f inches is %1.2f cenimeters" %(inches,inches_to_centimeter(inches)))

inches = -1.0
print("%1.2f inches is %1.2f cenimeters" %(inches,inches_to_centimeter(inches)))
````

````python
#
# have a single return at the end of the function 
#
def inches_to_centimeter(inch):
    if inch < 0: 
        result = 0
    else:
        result = inch*2.48 
    return result 

inches = 1.0
print("%1.2f inches is %1.2f cenimeters" %(inches,inches_to_centimeter(inches)))

inches = -1.0
print("%1.2f inches is %1.2f cenimeters" %(inches,inches_to_centimeter(inches)))
````

All the variables we define have scope. Scope is where the variable is accessable in the problem. By default, all variables are `local`. We can define variables to be `global`, this will make then accessable everywhere in the program. 

In our example function `concatinate_first_and_last_name(first_name,last_name)`, `first_name` and `last_name` are local variables. They are local to the function and only visible within the function. 

````python
global first_name 
global last_name 
concatinate_first_and_last_name()
````

### Function comments

When you write a function, you **should** also write down what it does. Good function comments explain:
 1. The purpose of the function
 2. The meaning of the parameters
 3. The meaning of the return value
 4. Any special requirements.
 
The format that we will use going forward is illustrated below. 

```
## Function purpose
# @param input variables and what they mean
# @return what does the return variable mean
#
def function_name(input_1(:type = default_value),input_2(:type = default_value), ... ):
    function implementation goes here
    return return_value(Default is None)
```

### Type hints

Type hints let us tell other people what the input and return data types are for the function. They are **optional** and provide similar look to strongly typed programming languages such as C/C++ and Java. Run the code below and observe that the output remained the same. 

````python
def concatinate_first_and_last_name(first_name: str,last_name: str):
    return first_name + " " + last_name

print(concatinate_first_and_last_name("Computer","Science"))
````

Python **DOES NOT** natively enforce or check varaible data types, you will need other tools like `mypi` to get this functionality.

### Default Values

Specifing default values is optional. However, the benifit is that your function will have a predefined behaviour if no input is provided. We will modify the above example so that the variable `x` has a default value of 1. 

````python
def concatinate_first_and_last_name(first_name: str = "Hello",last_name: str = "World"):
    return first_name + " " + last_name

print(concatinate_first_and_last_name("Memorial","University"))
print(concatinate_first_and_last_name())
````

Notice, that the second print statement is missing input into the function and there was no errors. Make note of the syntax if you want to include type hints in your code. The default value is specified after the data type.

### Keyword arguments

By default, function input parameters values are automatically set the in the order you write them. This is how we passed values to our function in the previous example. However, with keyword arguments you need to specify the parameter name and the value you want to assign it. As an added bonus, you can now set function parameters out of order. Run the code below and observe that you get the same out put in both cases.

````python
def concatinate_first_and_last_name(first_name: str = "Hellow",last_name: str = "World"):
    return first_name + " " + last_name

print(concatinate_first_and_last_name(first_name="Bell",last_name="Labs"))
print(concatinate_first_and_last_name(last_name="Labs",first_name="Bell"))
````
