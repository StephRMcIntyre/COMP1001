# Chapter 4b - For loops


In general, a `for` loop in python iterates over the contents any container. Some of Python's built-in containers are: dictionaires, lists, sets, and tuples. The `for` loop sytax is as follows,

```
    for loop variable in container:
        loop body code executed for every element in the container
    code after iterating over every element in the container 
```
every element in the container once and assigns it to the `loop variable`. To reiterate, the loop variable is an element of the container, not the index of the element in the container. As with `if` statements, the loop body code must be intened to be in the scope of the `for` loop.

### The range function

The range function is very useful, as it creates a sequence of integers. the `range` syntax is as follows,

```
    range(start,stop,step) 
```

The sequence you get from $range(start,stop,step)$ is (start,start+step,...,last), with $stop-step \leq last$ and $last < stop$. Note, $stop$ is not included. 

There are two special cases for range
- `range(start,stop)`, default value for step is 1
- `range(stop)`, default value for start is 0 and step is 1

The next three examples will illustrate the `range` function with a `for` loop. Run the example below and obeserve the output.

````python
for counter in range(5):
    print("In the loop counter =",counter)
print("Outside the loop counter =",counter)
````

`range(5)` creates the sequence (0,1,2,3,4) and `counter` is the loop variable. Notice, `counter` is assigned each value in the sequence once. After the loop terminates, the counter variable is still accessable.  We futher illustrate how the `for` loop assigns each element in `range(5)` to `counter` by tracing the code as follows,

### String processing application

A common `for` loops application is handling strings. For instance, you might want to count the number of vowels in a particular string. 

````python
string = "The swift brown fox jumped over the lazy dog"
vowels = 0
for char in string:
    if char in "aeiou":
        vowels = vowels + 1
        
print("The number of vowels in the string '%s' is %d  " %(string,vowels))
````

### For and While loops 

Recall, finding the start and stop value of a `while` was challenging. Do you use `<` or `<=` to terminate the loop? This is not the case with a `for` loop, we know start and stop values. The example  below is the `for` loop equivlent when `<` is the termination condition. 

````python
# Recall that < will execute stop - start number of times 
start = 0
stop = 5
print("The while loop:")
while start < stop:
    print("In the loop counter =",start)
    start = start + 1
print("Outside the loop counter =",start)

print("\nThe for loop:")
start = 0
for index in range(start,stop):
    print("In the loop counter =",index)
print("Outside the loop counter =",index)
````

The example below is the `for` loop equivlent when `<=` is the termination condition.

````python
# Recall that <= will execute stop - start+1 number of times 
start = 0
stop = 5
print("The while loop:")
while start <= stop:
    print("In the loop counter =",start)
    start = start + 1
print("Outside the loop counter =",start)

print("\nThe for loop:")
start = 0
for index in range(start,stop+1):
    print("In the loop counter =",index)
print("Outside the loop counter =",index)
````

### Nested loops

Whenever, you have a loop inside another loop you get a nested loops. You might use nested loops to  process data in a table. The first loop iterates over the rows of the table. The second loop iteratoes over the columns. For instance, we can print out the multiplation table using the code below. 

````python
for a in range(1,13):
    for b in range(1,10):
        print("%2d*%-2d = %3d" %(a,b,a*b),end=" | ")
    print("")
````

Notice, the inner loop is indented. As this is in the scope of the outer for loop. Moreover, the loop variable value for the outter loop is fixed until the inner loop is finished.

**Summary**

A `for` loop iterates through a container and assignes each element to a loop varaible. Every iteration assigns a different element in the container to the loop varaible. 

The `for` loop syntax is as follows:
 ```
     for loop variable in container:
         loop body code executed for every element in the container
     code after iterating over every element in the container 
 ```
Make sure to indent the code in the loop body. 
 
`range(start,stop,step)` generates the following integer sequence (start,start+step,...,last), with stop-step $\leq$ last and last $<$ stop.

A `while` loop using `start<stop` is the same as using `range(start,stop)` with a `for` loop.

A `while` loop using `start<=stop` is the same as using `range(start,stop+1)` with a `for` loop.

When using nested loops, the inner loop must be indented.
